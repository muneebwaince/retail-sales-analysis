```python
# ==================================================
# RETAIL SALES ANALYSIS PROJECT
# ==================================================

import pandas as pd
import matplotlib.pyplot as plt


# 1. LOAD DATA
df = pd.read_csv("retail_sales.csv")

pd.set_option("display.max_columns", None)

print("=" * 55)
print("1. DATA OVERVIEW")
print("=" * 55)

display(df.head())
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isna().sum())

print("\nDuplicate Rows:", df.duplicated().sum())


# 2. DATA VALIDATION
required_columns = [
    "Order_ID", "Order_Date", "City", "Category",
    "Product", "Quantity", "Unit_Price_PKR", "Unit_Cost_PKR"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

if df[required_columns].isna().any().any():
    raise ValueError(
        "Missing values found in required columns. "
        "Inspect and clean the data before continuing."
    )

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"], errors="coerce"
)

if df["Order_Date"].isna().any():
    raise ValueError("Invalid or missing dates found.")

if (df["Quantity"] <= 0).any():
    raise ValueError("Quantity must be greater than zero.")

if (df["Unit_Price_PKR"] < 0).any():
    raise ValueError("Unit price cannot be negative.")

if (df["Unit_Cost_PKR"] < 0).any():
    raise ValueError("Unit cost cannot be negative.")

print("\nData validation passed.")


# 3. DATA PREPARATION
df["Sales_PKR"] = (
    df["Quantity"] * df["Unit_Price_PKR"]
)

df["Total_Cost_PKR"] = (
    df["Quantity"] * df["Unit_Cost_PKR"]
)

df["Profit_PKR"] = (
    df["Sales_PKR"] - df["Total_Cost_PKR"]
)


# 4. OVERALL BUSINESS KPIs
total_sales = df["Sales_PKR"].sum()
total_cost = df["Total_Cost_PKR"].sum()
total_profit = df["Profit_PKR"].sum()

total_orders = df["Order_ID"].nunique()
total_units = df["Quantity"].sum()

aov = total_sales / total_orders if total_orders else 0

profit_margin = (
    total_profit / total_sales * 100
    if total_sales else 0
)

print("\n" + "=" * 55)
print("2. OVERALL BUSINESS SUMMARY")
print("=" * 55)

print(f"Total Sales: PKR {total_sales:,.2f}")
print(f"Total Cost: PKR {total_cost:,.2f}")
print(f"Total Profit: PKR {total_profit:,.2f}")
print(f"Overall Profit Margin: {profit_margin:.2f}%")
print(f"Total Unique Orders: {total_orders}")
print(f"Total Units Sold: {total_units}")
print(f"Average Order Value: PKR {aov:,.2f}")


# 5. CATEGORY-WISE ANALYSIS
category_summary = df.groupby("Category").agg(
    Sales_PKR=("Sales_PKR", "sum"),
    Total_Cost_PKR=("Total_Cost_PKR", "sum"),
    Profit_PKR=("Profit_PKR", "sum"),
    Units_Sold=("Quantity", "sum"),
    Orders=("Order_ID", "nunique")
)

category_summary["Profit_Margin_%"] = (
    category_summary["Profit_PKR"]
    / category_summary["Sales_PKR"].replace(0, float("nan"))
) * 100

category_summary["Profit_Margin_%"] = (
    category_summary["Profit_Margin_%"].round(2)
)

print("\n" + "=" * 55)
print("3. CATEGORY-WISE SUMMARY")
print("=" * 55)

display(category_summary)


# 6. MONTHLY SALES ANALYSIS
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Sales_PKR"]
    .sum()
    .sort_index()
)

monthly_summary = monthly_sales.to_frame(
    name="Sales_PKR"
)

monthly_summary["Previous_Month_Sales"] = (
    monthly_summary["Sales_PKR"].shift(1)
)

monthly_summary["Sales_Change_PKR"] = (
    monthly_summary["Sales_PKR"]
    - monthly_summary["Previous_Month_Sales"]
)

monthly_summary["Sales_Growth_%"] = (
    monthly_summary["Sales_Change_PKR"]
    / monthly_summary["Previous_Month_Sales"].replace(
        0, float("nan")
    )
) * 100

monthly_summary["Sales_Growth_%"] = (
    monthly_summary["Sales_Growth_%"].round(2)
)

print("\n" + "=" * 55)
print("4. MONTHLY SALES SUMMARY")
print("=" * 55)

display(monthly_summary)


# 7. CITY-WISE ANALYSIS
city_summary = df.groupby("City").agg(
    Sales_PKR=("Sales_PKR", "sum"),
    Total_Cost_PKR=("Total_Cost_PKR", "sum"),
    Profit_PKR=("Profit_PKR", "sum"),
    Units_Sold=("Quantity", "sum"),
    Orders=("Order_ID", "nunique")
)

city_summary["Profit_Margin_%"] = (
    city_summary["Profit_PKR"]
    / city_summary["Sales_PKR"].replace(0, float("nan"))
) * 100

city_summary["Profit_Margin_%"] = (
    city_summary["Profit_Margin_%"].round(2)
)

print("\n" + "=" * 55)
print("5. CITY-WISE SUMMARY")
print("=" * 55)

display(city_summary)


# 8. PRODUCT-WISE ANALYSIS
product_summary = df.groupby("Product").agg(
    Sales_PKR=("Sales_PKR", "sum"),
    Profit_PKR=("Profit_PKR", "sum"),
    Units_Sold=("Quantity", "sum")
).sort_values("Sales_PKR", ascending=False)

print("\n" + "=" * 55)
print("6. PRODUCT-WISE SUMMARY")
print("=" * 55)

display(product_summary)


# 9. BUSINESS INSIGHTS — AUTOMATIC
print("\n" + "=" * 55)
print("7. BUSINESS INSIGHTS")
print("=" * 55)

if not category_summary.empty:
    best_sales_category = category_summary["Sales_PKR"].idxmax()
    best_profit_category = category_summary["Profit_PKR"].idxmax()
    best_margin_category = category_summary["Profit_Margin_%"].idxmax()

    print(
        f"1. Highest Sales Category: {best_sales_category} "
        f"(PKR {category_summary.loc[best_sales_category, 'Sales_PKR']:,.2f})"
    )

    print(
        f"2. Highest Profit Category: {best_profit_category} "
        f"(PKR {category_summary.loc[best_profit_category, 'Profit_PKR']:,.2f})"
    )

    print(
        f"3. Highest Profit Margin Category: {best_margin_category} "
        f"({category_summary.loc[best_margin_category, 'Profit_Margin_%']:.2f}%)"
    )

if not city_summary.empty:
    best_sales_city = city_summary["Sales_PKR"].idxmax()
    best_profit_city = city_summary["Profit_PKR"].idxmax()
    best_margin_city = city_summary["Profit_Margin_%"].idxmax()

    print(
        f"4. Highest Sales City: {best_sales_city} "
        f"(PKR {city_summary.loc[best_sales_city, 'Sales_PKR']:,.2f})"
    )

    print(
        f"5. Highest Profit City: {best_profit_city} "
        f"(PKR {city_summary.loc[best_profit_city, 'Profit_PKR']:,.2f})"
    )

    print(
        f"6. Highest Profit Margin City: {best_margin_city} "
        f"({city_summary.loc[best_margin_city, 'Profit_Margin_%']:.2f}%)"
    )

if not monthly_sales.empty:
    best_sales_month = monthly_sales.idxmax()

    print(
        f"7. Highest Sales Month: {best_sales_month} "
        f"(PKR {monthly_sales.max():,.2f})"
    )

if not product_summary.empty:
    best_sales_product = product_summary["Sales_PKR"].idxmax()

    print(
        f"8. Highest Sales Product: {best_sales_product} "
        f"(PKR {product_summary.loc[best_sales_product, 'Sales_PKR']:,.2f})"
    )


# 10. CHART 1 — CATEGORY-WISE SALES AND PROFIT
category_summary[
    ["Sales_PKR", "Profit_PKR"]
].plot(kind="bar", figsize=(9, 5))

plt.title("Category-wise Sales and Profit")
plt.xlabel("Category")
plt.ylabel("Amount (PKR)")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBUSINESS INSIGHT — CATEGORY SALES AND PROFIT")
print(
    f"{best_profit_category} generated the highest total profit "
    f"at PKR {category_summary.loc[best_profit_category, 'Profit_PKR']:,.2f}."
)


# 11. CHART 2 — CATEGORY-WISE PROFIT MARGIN
category_summary["Profit_Margin_%"].plot(
    kind="bar",
    color="green",
    figsize=(8, 5)
)

plt.title("Category-wise Profit Margin")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBUSINESS INSIGHT — CATEGORY PROFIT MARGIN")
print(
    f"{best_margin_category} had the highest category profit margin "
    f"at {category_summary.loc[best_margin_category, 'Profit_Margin_%']:.2f}%."
)


# 12. CHART 3 — MONTHLY SALES TREND
monthly_chart = monthly_sales.copy()
monthly_chart.index = monthly_chart.index.astype(str)

monthly_chart.plot(
    kind="line",
    marker="o",
    figsize=(8, 5)
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (PKR)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBUSINESS INSIGHT — MONTHLY SALES")
print(
    f"{best_sales_month} recorded the highest monthly sales "
    f"at PKR {monthly_sales.max():,.2f}."
)

if len(monthly_sales) >= 2:
    first_month_sales = monthly_sales.iloc[0]
    second_month_sales = monthly_sales.iloc[1]

    if first_month_sales != 0:
        growth = (
            (second_month_sales - first_month_sales)
            / first_month_sales
        ) * 100

        print(
            f"Sales changed by {growth:.2f}% from "
            f"{monthly_sales.index[0]} to {monthly_sales.index[1]}."
        )


# 13. CHART 4 — CITY-WISE PROFIT
city_summary["Profit_PKR"].plot(
    kind="bar",
    color="green",
    figsize=(8, 5)
)

plt.title("City-wise Profit")
plt.xlabel("City")
plt.ylabel("Profit (PKR)")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBUSINESS INSIGHT — CITY PROFIT")
print(
    f"{best_profit_city} generated the highest total profit "
    f"at PKR {city_summary.loc[best_profit_city, 'Profit_PKR']:,.2f}."
)


# 14. CHART 5 — CITY-WISE PROFIT MARGIN
city_summary["Profit_Margin_%"].plot(
    kind="bar",
    color="orange",
    figsize=(8, 5)
)

plt.title("City-wise Profit Margin")
plt.xlabel("City")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBUSINESS INSIGHT — CITY PROFIT MARGIN")
print(
    f"{best_margin_city} had the highest city profit margin "
    f"at {city_summary.loc[best_margin_city, 'Profit_Margin_%']:.2f}%."
)


# 15. CHART 6 — PRODUCT-WISE SALES
product_summary["Sales_PKR"].sort_values().plot(
    kind="barh",
    color="steelblue",
    figsize=(9, 5)
)

plt.title("Product-wise Sales")
plt.xlabel("Sales (PKR)")
plt.ylabel("Product")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBUSINESS INSIGHT — PRODUCT SALES")
print(
    f"{best_sales_product} generated the highest product sales "
    f"at PKR {product_summary.loc[best_sales_product, 'Sales_PKR']:,.2f}."
)


# 16. PROJECT CONCLUSION
print("\n" + "=" * 55)
print("8. PROJECT CONCLUSION")
print("=" * 55)

print(
    f"The dataset contains {total_orders} unique orders "
    f"and {total_units} units sold."
)

print(
    f"Total sales were PKR {total_sales:,.2f}, "
    f"with total profit of PKR {total_profit:,.2f} "
    f"and an overall profit margin of {profit_margin:.2f}%."
)

if not category_summary.empty and not city_summary.empty:
    print(
        f"{best_sales_category} led category sales, while "
        f"{best_margin_category} had the highest category profit margin."
    )

    print(
        f"{best_profit_city} led city profit, while "
        f"{best_margin_city} had the highest city profit margin."
    )


# 17. PROJECT LIMITATIONS
print("\n" + "=" * 55)
print("9. PROJECT LIMITATIONS")
print("=" * 55)

print("- Results depend on the quality and completeness of the dataset.")
print("- The dataset covers only the dates included in the CSV file.")
print("- Profit uses recorded unit cost and does not include other expenses")
print("  such as rent, salaries, delivery, taxes or marketing.")
print("- Findings describe this dataset and may not represent a real business.")


# 18. FINAL DATASET PREVIEW
print("\n" + "=" * 55)
print("10. FINAL DATASET")
print("=" * 55)

display(df.head())


# 19. OPTIONAL — INTERACTIVE GOOGLE SHEET
# Run in Google Colab only.

from google.colab import sheets

sheet = sheets.InteractiveSheet(df=df)
```
